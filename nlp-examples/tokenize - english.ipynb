```python
from nltk import sent_tokenize, word_tokenize, wordpunct_tokenize, pos_tag
# pos = part of speech 문법도 본다. 
```

```python
text = "The old building is scheduled for demolition. The contractors will begin building a new structure next month."
```

```python
text = "I am actively looking for Ph.D. students. and you are a Ph.D student."
```

```python
for sent in sent_tokenize(text):
    print(sent)
```

```python
for sent in sent_tokenize(text):
    for word in word_tokenize(sent):
    # for word in wordpunct_tokenize(sent):
        print(word)

    print()
```
