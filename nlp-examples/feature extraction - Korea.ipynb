```python
# The corpus object
corpus = [
    "반얀트리 서울, 스포츠 시설 리뉴얼 기념 농구 대회 개최",
    "금융위에 P2P업체 투게더펀딩·펀다 온투업 등록 신청",
    "문 닫힌 롯데백화점 본점",
]
```

```python
# regex: regular expression, 따로 라도 꼭 공부하는게 좋다.중요. 모든 텍스트는 정규화로 다룬다.
import re 
```

```python
def clean_doc(doc):
    # regular expression - subsitution(replace)
    # \w : word character (a-z, A-Z, 0-9, _, ...)
    # \s : space
    # [^...] : excluding ...
    doc = re.sub(r'[^\w\s]', ' ', doc) # doc에서, 앞에있는표현을, 뒤에있는 것으로 바꿔줘.[abc] a또는b또는c .. 중에 하나. a-z a부터z, a-zA-Z 대문자까지. \w 워드 케릭터 \s 공간 엔터 등 
       #... 결국, 특수문자 제거해줘.
    return doc
```

```python
corpus = [clean_doc(x) for x in corpus]
```

## Compare Tokenizers

```python
from konlpy.tag import Okt 
#konlpy 한국어 공통 처리 인터페이스.. 다양한 모듈 제공해요. 
```

```python
okt = Okt()
```

## Feature Extraction

```python
import gensim 
```

```python
tokenized_corpus = [
    han.morphs(doc) for doc in corpus
    
]
```

```python
tokenized_corpus
```

```python
lexicon = gensim.corpora.Dictionary(tokenized_corpus)
```

```python
for x in lexicon.items():
    print(x)
```

### (1) Bag of words(BoW)

```python
for doc in tokenized_corpus:
    vec = lexicon.doc2bow(doc)
    print(vec)
```

### (2) One - Hot Encoding

```python
for doc in tokenized_corpus:
    vec = lexicon.doc2bow(doc)
    vec = [(x[0], 1) for x in vec] #추가하기
    print(vec)
```

### (3) Tf-Idf

```python
tfidf = gensim.models.TfidfModel(dictionary=lexicon, normalize = True)
```

```python
for doc in tokenized_corpus:
    vec = lexicon.doc2bow(doc)
    vec = tfidf[vec] #가중치 적용하기. 
    print(vec)
    
```
