## Bag of Words 

### (1) nltk & gensim

```python
import nltk
import string
```

```python
# stemmer / lematize
def tokenize(text):
    text = text.lower()
    stemmer = nltk.stem.SnowballStemmer('english')


    for token in nltk.word_tokenize(text):
        if token in string.punctuation:
            continue

        yield stemmer.stem(token)
        
```

```python
for doc in corpus:
    print(doc)

    tokens = list(tokenize(doc))
    print(tokens)
    print()
```

```python
corpus
```

```python
tokenized_corpus = [
    list(tokenize(doc)) for doc in corpus
]
```

```python
tokenized_corpus #기계한테 넘기기 전에 전처리 한다. 
```

```python
import gensim
```

```python
lexicon = gensim.corpora.Dictionary(tokenized_corpus)
```

```python
len(lexicon)
```

```python
for x in lexicon.items():
    print(x)
```

```python
for doc in tokenized_corpus:
    print(doc)

    #document to baf-of-words
    vec = lexicon.doc2bow(doc)

    print(vec)
    print()
```

```python
# lazy evaluation (일시정지) - 필요할 때 하나 씩 꺼낸다. - 메모리를 아 낄 수 있 다. 
def func2():
    for i in range(3):
        print(i)  # 추가 해보자
        yield i # 뭐지뭐지뭐지뭐지뭐지뭐지뭐지뭐지
```

```python
for x in func2():
    print(x)
```

## 3. Tf-Idf

$$
\begin{align}
    tf (t, d) &= 1 + \log \, f_{t,d}  \\
    idf (t,D) &= \log 1 + \frac{N}{n_t} \\
    tf-idf (t, d, D) &= tf (t,d) \cdot idf (t,D)
\end{align}
$$

# (1) Gensim 

```python
# Gensim -> 강조 이유 한글 데이터 사용할거라서.. 
```

```python
tfidf = gensim.models.TfidfModel(dictionary=lexicon, normalize = True)
```

```python
#Tf-idf : 중요하지 않은 단어(모든 문장에 들어가는 단어) 부정관사, 전치사... 등 결국 그 단어의 의미와는 영향 없다.
     #     중요한 단어. 일부 문장에만 들어간다. 가중치를 올려준다. 
```

```python
for doc in tokenized_corpus:
    print(doc)

    vec = lexicon.doc2bow(doc)
    vec = tfidf[vec]

    print(vec)
    print()
```
