```python
import kss
```

```python
text = '너에게 닥친 고난이 아무리 커보이더라도 명심해라. 아직 빙산의 일각일 뿐이다.'
```

```python
for sent in kss.split_sentences(text):
    print(sent)
```

```python
# pip install konlpy
```

```python
from konlpy.tag import Okt
```

```python
# instance <- class
okt = Okt()
```

```python
for sent in kss.split_sentences(text):
    print(sent)

    for token in okt.pos(sent):
        print(token)

    print()
```
